```python
#loading libraries:
from tensorflow.keras.layers import Dense,Conv2D,BatchNormalization,Dropout,Flatten
from tensorflow.keras.models import Sequential
import keras
import tensorflow
from matplotlib import pyplot as plt
from tensorflow.keras.callbacks import EarlyStopping
```

```python
#data generator: gives images in batches:
train_data = keras.utils.image_dataset_from_directory(
    directory = r"C:\Users\Anushkha Bhagra\Downloads\deep learning\CNN\data\train",
    image_size = (256,256),
    labels = "inferred",
    label_mode = "int",
    batch_size = 32
)

validation_ds = keras.utils.image_dataset_from_directory(
    directory = r"C:\Users\Anushkha Bhagra\Downloads\deep learning\CNN\data\val",
    image_size  = (256,256),
    labels = "inferred",
    label_mode = "int",
    batch_size = 32
)
```

```python
#data normalization: value of every image between 0 and 1
def process(image,label):
    image = tensorflow.cast(image/255.,tensorflow.float32) #cast changing the type of data
    return image,label
train_data = train_data.map(process)
validation_ds = validation_ds.map(process)
```

```python
#cnn flow:
model = Sequential()

#input layer:
model.add(Conv2D(32,kernel_size=(3,3),activation = "relu", input_shape = (256,256,3)))

#hidden layer:
model.add(Conv2D(64,kernel_size=(3,3),activation = "relu"))

model.add(Conv2D(128,kernel_size=(3,3),activation = "relu"))

#convert 2d into 1d:
model.add(Flatten())

#dense layer:
model.add(Dense(128,activation ="relu"))

#output layer:'
model.add(Dense(1,activation="sigmoid"))

model.summary()
```

```python
# compile:
model.compile(optimizer = "adam", loss = "binary_crossentropy", metrics = ["accuracy"])
```

```python
#model trainig:
model.fit(train_data,epochs = 10, batch_size = 32, validation_data = validation_ds,verbose = 1)
```
