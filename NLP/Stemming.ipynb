```python
#stemming:
words = ["playing", "shocking", "shocked", "played", "going", "gone", "history"]
```

```python
from nltk.stem import PorterStemmer

stemming = PorterStemmer()
for i in words:
    print(i, "-->", stemming.stem(i))
```

```python
#2nd method:
from nltk.stem import SnowballStemmer  #also known as porter 2

stemming = SnowballStemmer("english")
for i in words:
    print(i, "-->", stemming.stem(i))
```
