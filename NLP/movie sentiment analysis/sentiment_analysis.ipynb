```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\deep learning\NLP\movie sentiment analysis\IMDB Dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
label = LabelEncoder()
df["encoded_sentiment"] = label.fit_transform(df["sentiment"])
var = dict(zip(df["sentiment"], df["encoded_sentiment"]))
print(var)
```

```python
df.drop(columns=["sentiment"],inplace = True)
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words("english"))
def preprocess(text):
    text = str(text)
    text = text.lower()
    words = text.split()
    words = [lemmatizer.lemmatize(word, pos = "v") for word in words if word  not in stop_words]
    return " ".join(words)
```

```python
df["reviews"] = df["review"].apply(preprocess)
```

```python
#feature selection:
x = df["reviews"]
y = df["encoded_sentiment"]
```

```python
#data splitting:
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state = 42, stratify = y)
```

```python
#vectorization:
vector = CountVectorizer(max_features = 500, binary = True)
x_train_vector = vector.fit_transform(x_train)
x_test_vector = vector.transform(x_test)
```

```python
#model selection:
model = LogisticRegression(max_iter = 10000)
model.fit(x_train_vector,y_train)
```

```python
y_pred = model.predict(x_test_vector)
accuracy_score(y_test,y_pred)
```

```python
y_pred_train = model.predict(x_train_vector)
accuracy_score(y_train,y_pred_train)
```
