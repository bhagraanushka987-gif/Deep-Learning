```python
import nltk
nltk.download('wordnet')
```

```python
#stemming:
words = ["playing", "shocking", "shocked", "played", "going", "gone", "history","pulls", "pulled","pulling"]
```

```python
from nltk.stem import WordNetLemmatizer
lemmetizor = WordNetLemmatizer()
for i in words:
    print(i, "-->", lemmetizor.lemmatize(i,pos="v"))
```
